La base de datos externa simula cancelaciones de membresías de **GymPulse**, una cadena de gimnasios similar a FitBalance Gym. **Pregunta del análisis:** ¿Qué peso debería asignar FitBalance a su información propia y qué peso podría recibir la información externa de GymPulse al estimar el riesgo económico?

Se seguirá el orden del notebook de referencia `RETO_3_CREDIBILIDAD`: primero el error de los datos internos y después la EPV, la media hipotética, la VHM y el coeficiente de credibilidad. Se agregan explicaciones para distinguir las fórmulas del ejercicio de sus interpretaciones estadísticas.

## 1. Carga de datos internos y caracterización del riesgo

La base observada corresponde a **FitBalance Gym**. Cada fila representa el registro de un día y contiene una frecuencia y una severidad.

- **Frecuencia:** número de cancelaciones registradas.
- **Severidad:** pérdida económica promedio por cliente, en COP.
- **LDA:** pérdida agregada del registro, calculada como frecuencia × severidad.
- **ND:** cantidad de registros válidos utilizados.
- **u:** media de las pérdidas agregadas.
- **sigma:** desviación estándar, que mide la dispersión de las pérdidas.

Se utiliza la misma hoja `Datos` y las mismas columnas 5 y 6 (índices de Python) del Reto 1. No se crea ni se sustituye ningún dataset.

```python
import numpy as np
import pandas as pd

# 1. Cargamos los datos internos de FitBalance
path = "/content/drive/MyDrive/Integracion de datos/Reto_1_Clientes_Gimnasio.xlsx"

XDB = pd.read_excel(path, sheet_name="Datos")
display(XDB.head())

Freq = pd.to_numeric(XDB.iloc[:, 5], errors="coerce")
Sev = pd.to_numeric(XDB.iloc[:, 6], errors="coerce")
LDA = (Freq * Sev).dropna()
ND = len(LDA)

# 2. Media y desviación estándar de las pérdidas
u = np.mean(LDA)
sigma = np.std(LDA)

print("Número de datos internos:", ND)
print("Media de las pérdidas:", u, "COP")
print("Desviación estándar:", sigma, "COP")
```

## 2. Estimación del intervalo y del error

### Intervalo de los datos: media ± 2 desviaciones estándar

Siguiendo la referencia de clase, se calcula `u - 2*sigma` y `u + 2*sigma`. Este intervalo describe la dispersión aproximada de los valores alrededor de la media. **No es un intervalo de confianza de la media**. Bajo una distribución normal, aproximadamente el 95% de los valores se encuentra dentro de dos desviaciones estándar; para datos de pérdidas, la aproximación puede no cumplirse exactamente.

### Error con la fórmula de proporciones del ejercicio

La referencia utiliza:

\[
E=z\sqrt{\frac{P(1-P)}{ND}}
\]

Donde `z = 2`, `P = 0.95` y `ND` es el número de observaciones. Se conserva esta fórmula para reproducir la metodología del ejercicio. Sin embargo, **P debe representar una proporción o probabilidad que se está estimando**: asumir que P es 0.95 no convierte automáticamente el resultado en el error monetario de las pérdidas. Por esa razón, el resultado de E se presenta por separado y no se suma a una media expresada en COP.

```python
# Se estima el intervalo de los datos según la referencia
print("El límite inferior de los datos es:", u - 2*sigma, "COP")
print("El límite superior de los datos es:", u + 2*sigma, "COP")

# Error de proporciones según la fórmula trabajada en clase
z = 2
P = 0.95

E = z * np.sqrt((P * (1-P)) / ND)
print("El error calculado con la fórmula de proporciones es:", E)
print("El error expresado en puntos porcentuales es:", E*100)
```

### Intervalo de confianza de la media de pérdidas en COP

Para responder cuánto podría variar la **media monetaria estimada**, se utiliza la fórmula del error estándar de la media:

\[
E_{\text{media}}=1.96\frac{s}{\sqrt{ND}}
\]

Aquí `s` es la desviación estándar muestral. Este margen sí tiene unidades de COP, por lo que puede sumarse y restarse a la media. El intervalo es una aproximación normal al 95% y supone que las observaciones son suficientemente independientes y representativas. No es una garantía de cobertura cuando existe dependencia temporal o una distribución muy asimétrica.

A diferencia de la referencia portuaria, no se multiplica por 1.000: la severidad de FitBalance ya está expresada en COP.

```python
# Margen de error de la media monetaria (aproximación normal al 95%)
s = np.std(LDA, ddof=1)
E_media = 1.96 * s / np.sqrt(ND)

print("El intervalo de la media con el error es:")
print("La media menos el error:", u - E_media, "COP")
print("La media más el error:", u + E_media, "COP")
print("El desfase total del intervalo es:", 2*E_media, "COP")
print("El margen de error relativo a la media es:", (E_media/u)*100, "%")
```

## 3. Carga de datos observados y datos externos

En esta parte se prepara la información que se utilizará para calcular la credibilidad. Se mantienen dos bases independientes:

|---|---|---|
| Observada | FitBalance Gym | Información propia cuya estimación se desea complementar. |

Ambas bases se expresan en COP y se comparan mediante la misma variable de riesgo: pérdida agregada por registro. La comparación supone que las definiciones de frecuencia, severidad y período de observación son suficientemente equivalentes. La credibilidad estadística no sustituye la validación de calidad, procedencia y compatibilidad de los datos.

```python
# 1. Cargamos los datos observados de FitBalance
# Se conserva el mismo archivo utilizado en el Reto 1
nxl = "/content/drive/MyDrive/Integracion de datos/Reto_1_Clientes_Gimnasio.xlsx"

XDB = pd.read_excel(nxl, sheet_name="Datos")

Freq = pd.to_numeric(XDB.iloc[:, 5], errors="coerce")
Sev = pd.to_numeric(XDB.iloc[:, 6], errors="coerce")
LDA = (Freq * Sev).dropna()

ND = len(LDA)
sigmao = np.std(LDA)
uo = np.mean(LDA)

print("BASE OBSERVADA - FITBALANCE")
print("Número de datos:", ND)
print("Media observada:", uo, "COP")
print("Desviación estándar observada:", sigmao, "COP")
```

```python
# 2. Cargamos los datos externos de GymPulse
nxle = "/content/drive/MyDrive/Integracion de datos/Base_de_Datos_GymPulse_Externa.xlsx"

XDE = pd.read_excel(nxle, sheet_name=0)

# Columnas originales: frecuencia de cancelaciones y severidad por cliente
Freqe = pd.to_numeric(XDE["Frecuencia_Cancelaciones_Dia"], errors="coerce")
Seve = pd.to_numeric(XDE["Severidad_COP_Cliente"], errors="coerce")
LDAe = (Freqe * Seve).dropna()

NDe = len(LDAe)
sigmae = np.std(LDAe)
ue = np.mean(LDAe)

print("BASE EXTERNA - GYMPULSE")
print("Número de datos:", NDe)
print("Media externa:", ue, "COP")
print("Desviación estándar externa:", sigmae, "COP")
```

## 4. Cálculo de la teoría de credibilidad

Se reproducen las tres medidas del ejemplo de clase, con las variables originales.

**EPV — Varianza ponderada.** Combina la variabilidad de las pérdidas dentro de cada base. Una EPV elevada significa que los registros individuales presentan una dispersión considerable.

\[
EPV=\frac{ND\,\sigma_o^2+NDe\,\sigma_e^2}{ND+NDe}
\]

**Media hipotética (uh).** Es el promedio de las dos medias, ponderado por el número de registros. Representa el nivel de riesgo promedio del conjunto de referencia.

\[
uh=\frac{ND\,uo+NDe\,ue}{ND+NDe}
\]

**VHM — Varianza de las medias hipotéticas.** Mide cuánto difieren las medias de FitBalance y GymPulse respecto a la media ponderada. Una VHM cercana a cero indica que sus medias son muy similares.

\[
VHM=\frac{ND\,uo^2+NDe\,ue^2}{ND+NDe}-uh^2
\]

**Cr — Coeficiente de credibilidad.** La fórmula utilizada en el ejercicio relaciona el número de datos internos con la razón EPV/VHM:

\[
Cr=\frac{ND}{ND+EPV/VHM}
\]

El resultado se encuentra entre 0 y 1. En la interpretación de la referencia, `Cr` es el peso de los datos propios y `1-Cr` es el peso complementario asignado a la información de referencia. Un coeficiente alto no significa que una base sea verdadera en ese mismo porcentaje.

```python
# 1. Varianza ponderada
EPV = (ND*sigmao**2 + NDe*sigmae**2) / (ND+NDe)
print("La ponderación de las varianzas (EPV) es:", EPV)

# 2. Media hipotética - ponderado de las medias
uh = (ND*uo + NDe*ue) / (ND+NDe)
print("La media ponderada de los datos es:", uh)

# 3. Varianza de las medias hipotéticas
VHM = (
    (ND/(ND+NDe))*uo**2
    + (NDe/(ND+NDe))*ue**2
    - uh**2
)
print("La varianza de las medias hipotéticas (VHM) es:", VHM)

# 4. Credibilidad según la fórmula del ejercicio
if VHM > 0:
    Cr = ND / (ND + (EPV/VHM))
    print("La credibilidad de mis datos es del:", Cr*100, "%")
    print("La credibilidad de los datos externos es del:", (1-Cr)*100, "%")
else:
    Cr = np.nan
    print("No se puede calcular el coeficiente con esta fórmula porque VHM es cero.")
```

### ¿Cómo se interpreta el coeficiente?

El cálculo anterior estima cuánto peso debe conservar la información interna en el modelo simplificado de la referencia. El complemento `1-Cr` no es una auditoría de calidad de GymPulse ni una probabilidad de que sus datos sean correctos.

En la formulación de credibilidad tipo Bühlmann, la combinación se realiza entre la media propia y la media hipotética:

\[
u_{cred}=Cr\,uo+(1-Cr)\,uh
\]

Por tanto, el complemento pesa sobre `uh`, que incluye información de ambas bases. No debe confundirse automáticamente con el peso directo de la media de GymPulse. Se añade este cálculo únicamente para mostrar cómo se aplicaría el coeficiente obtenido.

```python
# Estimación ajustada mediante el coeficiente de credibilidad
if not np.isnan(Cr):
    u_cred = Cr*uo + (1-Cr)*uh
    print("Media propia de FitBalance:", uo, "COP")
    print("Media hipotética ponderada:", uh, "COP")
    print("Media ajustada por credibilidad:", u_cred, "COP")
    print("Diferencia frente a la media propia:", u_cred-uo, "COP")
```
