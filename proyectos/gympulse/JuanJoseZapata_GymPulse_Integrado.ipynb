FitBalance Gym es una cadena ficticia de gimnasios que busca analizar el riesgo económico relacionado con la cancelación de membresías.

Frecuencia: indica el número de clientes que cancelan su membresía diariamente (clientes/día).

Se procede con la construcción del histograma de frecuencias

```python
NI = 10 # Número de intervalos
counts, Limits = np.histogram(LDA, bins=NI) # Agrupar los datos
LI = Limits[:-1] # Límite inferior
LS = Limits[1:] # Límite superior
fi = counts / sum(counts) # Porcentaje de datos por intervalo
MC = (LI + LS) / 2 # Marca de clase

histogram_data = pd.DataFrame({
    'LI': LI,
    'LS': LS,
    'ND': counts,
    'fi': fi,
    'MC': MC
})
display(histogram_data)
```

Medidas de tendencia central y dispersión

```python
np.set_printoptions(suppress=True)

# Media
U = np.sum(MC * fi)
print(f"La media de las pérdidas es: {U:.2f}")

# Varianza
varc = np.sum((MC - U)**2 * fi)
print(f"La varianza es: {varc:.2f}")

# Intervalo de variación de las pérdidas
sigma = np.sqrt(varc)
print(f"El límite inferior es: {U - 2*sigma:.2f}")
print(f"El límite superior es: {U + 2*sigma:.2f}")

# Coeficiente de asimetría
cas = np.sum(((MC - U)**3) * fi) / (sigma**3)
print(f"El coeficiente de asimetría es: {cas:.2f}")

# Coeficiente de curtosis
cur = np.sum(((MC - U)**4) * fi) / (sigma**4)
print(f"El coeficiente de curtosis es: {cur:.2f}")
```

## 5. Medidas solicitadas

```python
media = np.sum(MC * fi)
varianza = np.sum((MC - media)**2 * fi)
desviacion = np.sqrt(varianza)
asimetria = np.sum((MC - media)**3 * fi) / desviacion**3
curtosis = np.sum((MC - media)**4 * fi) / desviacion**4

resultados = pd.DataFrame({
    "Medida": ["Media", "Varianza", "Desviación estándar", "Asimetría", "Curtosis"],
    "Resultado": [media, varianza, desviacion, asimetria, curtosis]
})

resultados
```

```python
def MedidasDTC(Xi, NI=10):
    """Calcula las medidas pedidas por el reto usando el método de datos agrupados visto en clase."""
    Xi = pd.Series(Xi).dropna().to_numpy(dtype=float)

    counts, Limits = np.histogram(Xi, bins=NI)
    LI = Limits[:-1]
    LS = Limits[1:]
    fi = counts / counts.sum()
    MC = (LI + LS) / 2

    media = np.sum(MC * fi)
    varianza = np.sum((MC - media)**2 * fi)
    desviacion = np.sqrt(varianza)
    asimetria = np.sum((MC - media)**3 * fi) / desviacion**3
    curtosis = np.sum((MC - media)**4 * fi) / desviacion**4

    tabla_hist = pd.DataFrame({
        'Limite Inferior': LI,
        'Limite Superior': LS,
        'ND': counts,
        'fi': fi,
        'MC': MC
    })

    medidas = pd.Series({
        'Media': media,
        'Varianza': varianza,
        'Desviación estándar': desviacion,
        'Asimetría': asimetria,
        'Curtosis': curtosis
    })

    return tabla_hist, medidas
```

## 6. Reto 2 - Muestreo aleatorio siguiendo el procedimiento trabajado en clase

Para esta parte se mantiene la lógica del ejercicio visto en clase:

1. Se divide la LDA en cajas o intervalos.
2. Se calcula la proporción de datos que cae en cada caja.
3. Esa proporción se convierte en cantidad de datos a seleccionar.
4. Se utiliza `random.choices()` para construir una nueva muestra.
5. Se vuelven a calcular Media, Varianza, Asimetría y Curtosis.
6. Se comparan las medidas antes y después del muestreo.

Se realizan dos tamaños de muestra: 1.000 y 10.000 datos, de acuerdo con los niveles solicitados en el reto.


```python
import random

# Limpiar la distribución agregada de pérdidas
LDA_original = pd.Series(LDA).dropna().to_numpy(dtype=float)

print("Cantidad de datos originales:", len(LDA_original))
```

### 6.1 Construcción de las cajas


```python
# En el ejercicio de clase se trabaja con cajas.
# Para mantener esa lógica se utilizarán 5 intervalos.

NC = 5

counts_box, limits_box = np.histogram(LDA_original, bins=NC)

Boxes = []

for j in range(NC):
    if j == NC - 1:
        caja = LDA_original[
            (LDA_original >= limits_box[j]) &
            (LDA_original <= limits_box[j + 1])
        ]
    else:
        caja = LDA_original[
            (LDA_original >= limits_box[j]) &
            (LDA_original < limits_box[j + 1])
        ]

    Boxes.append(list(caja))

# Probabilidad o porcentaje de datos por caja
pm = counts_box / np.sum(counts_box)

tabla_cajas = pd.DataFrame({
    "Caja": np.arange(1, NC + 1),
    "Límite inferior": limits_box[:-1],
    "Límite superior": limits_box[1:],
    "Número de datos": counts_box,
    "Proporción": pm
})

display(tabla_cajas)
```

### 6.2 Medidas antes del muestreo


```python
hist_original, medidas_original = MedidasDTC(LDA_original, NI=10)

print("MEDIDAS ANTES DEL MUESTREO")
display(medidas_original.to_frame("Original"))
```

### 6.3 Primera muestra: 1.000 datos


```python
# Convertir las proporciones en cantidades para una muestra de 1.000 datos
pm_1000 = (np.array(pm) * 1000).astype(int)

# Ajustar por redondeo para obtener exactamente 1.000 datos
pm_1000[-1] = pm_1000[-1] + (1000 - np.sum(pm_1000))

print("Número de datos a sacar de cada caja:")
print(pm_1000)
print("Total:", np.sum(pm_1000))

# Construcción de la nueva LDA siguiendo el procedimiento de clase
LDA2 = []

random.seed(123)

for j in range(NC):
    if pm_1000[j] > 0 and len(Boxes[j]) > 0:
        LDA2.extend(random.choices(Boxes[j], k=pm_1000[j]))

LDA2 = np.array(LDA2)

print("Cantidad de datos de LDA2:", len(LDA2))
```

```python
# Medidas después del primer muestreo
hist_1000, medidas_1000 = MedidasDTC(LDA2, NI=10)

print("MEDIDAS DESPUÉS DEL MUESTREO - 1.000 DATOS")
display(medidas_1000.to_frame("Muestra 1.000"))
```

### 6.4 Segunda muestra: 10.000 datos


```python
# Convertir las proporciones en cantidades para una muestra de 10.000 datos
pm_10000 = (np.array(pm) * 10000).astype(int)

# Ajustar por redondeo para obtener exactamente 10.000 datos
pm_10000[-1] = pm_10000[-1] + (10000 - np.sum(pm_10000))

print("Número de datos a sacar de cada caja:")
print(pm_10000)
print("Total:", np.sum(pm_10000))

# Construcción de la segunda muestra
LDA3 = []

random.seed(456)

for j in range(NC):
    if pm_10000[j] > 0 and len(Boxes[j]) > 0:
        LDA3.extend(random.choices(Boxes[j], k=pm_10000[j]))

LDA3 = np.array(LDA3)

print("Cantidad de datos de LDA3:", len(LDA3))
```

```python
# Medidas después del segundo muestreo
hist_10000, medidas_10000 = MedidasDTC(LDA3, NI=10)

print("MEDIDAS DESPUÉS DEL MUESTREO - 10.000 DATOS")
display(medidas_10000.to_frame("Muestra 10.000"))
```

## 7. Comparación antes y después


```python
comparacion = pd.concat([
    medidas_original.rename("Original"),
    medidas_1000.rename("Muestra 1.000"),
    medidas_10000.rename("Muestra 10.000")
], axis=1)

display(comparacion)
```

## 8. Diferencia porcentual respecto a los datos originales


```python
comparacion_error = comparacion.copy()

comparacion_error["Diferencia % - 1.000"] = (
    (comparacion_error["Muestra 1.000"] - comparacion_error["Original"])
    / comparacion_error["Original"]
) * 100

comparacion_error["Diferencia % - 10.000"] = (
    (comparacion_error["Muestra 10.000"] - comparacion_error["Original"])
    / comparacion_error["Original"]
) * 100

display(comparacion_error)
```

# 10. Integración de datos externos - GymPulse

En esta parte FitBalance Gym integra la base de datos externa de **GymPulse**, siguiendo la misma lógica utilizada en el ejercicio de integración con COSCO Shipping.

El procedimiento consiste en:

1. Construir cinco concentradores de información con los datos internos.
2. Clasificar los datos internos en cinco cajas.
3. Obtener una muestra base de 1.000 datos.
4. Cargar la información externa de GymPulse.
5. Calcular el valor de pertenencia de cada dato externo frente a los concentradores internos.
6. Integrar únicamente los datos externos con un valor de pertenencia superior al 90%.
7. Volver a realizar el muestreo después de la integración.
8. Comparar la distribución original, la muestra interna y la muestra con datos externos.


## 10.1 Construcción de concentradores y cajas con los datos internos


```python
import random
import numpy as np

# Utilizar la LDA original de FitBalance
LDA_int = np.asarray(pd.Series(LDA).dropna(), dtype=float)

# Crear cinco concentradores de información
random.seed(42)
XC_int = np.array(random.choices(list(LDA_int), k=5), dtype=float)
XC_int = np.sort(XC_int)

print("Los concentradores iniciales de información son:")
print(XC_int)

# Crear las cinco cajas
Boxes_int = [[], [], [], [], []]

# Clasificar los datos internos según el concentrador más cercano
for k in range(len(LDA_int)):

    # Evitar división entre cero
    denominador = np.where(np.abs(XC_int) < 1e-12, 1e-12, np.abs(XC_int))

    # Distancia relativa del dato a cada concentrador
    d = np.abs(XC_int - LDA_int[k]) / denominador

    # Identificar caja
    nc = np.argmin(d)

    # Actualizar el centroide
    XC_int[nc] = (XC_int[nc] + LDA_int[k]) / 2

    # Guardar el dato
    Boxes_int[nc].append(LDA_int[k])

print("Concentradores actualizados:")
print(XC_int)

print("Cantidad de datos por caja:")
print([len(caja) for caja in Boxes_int])
```

## 10.2 Porcentaje de datos por caja y muestreo interno


```python
# Porcentaje de datos por caja
pm_int = []

for j in range(5):
    porcentaje = len(Boxes_int[j]) / len(LDA_int)
    pm_int.append(porcentaje)

# Convertir los porcentajes a cantidad de datos para una muestra de 1.000
pm_int = (np.array(pm_int) * 1000).astype(int)

# Ajustar por redondeo para completar exactamente 1.000 datos
pm_int[-1] = pm_int[-1] + (1000 - np.sum(pm_int))

print("Número de datos que se tomarán de cada caja:")
print(pm_int)
print("Total:", np.sum(pm_int))

# Crear muestra interna antes de integrar datos externos
LDA_muestra_interna = []

random.seed(123)

for j in range(5):
    LDA_muestra_interna.extend(
        random.choices(Boxes_int[j], k=pm_int[j])
    )

LDA_muestra_interna = np.array(LDA_muestra_interna)

print("Cantidad de datos de la muestra interna:", len(LDA_muestra_interna))
```

## 10.3 Carga y caracterización de la base externa GymPulse


```python
# Ruta de la base externa GymPulse
path_gympulse = "/content/drive/MyDrive/Integracion de datos/Base_de_Datos_GymPulse_Externa.xlsx"

# Cargar base
GymPulse = pd.read_excel(path_gympulse, sheet_name=0)

display(GymPulse.head())

# Frecuencia y severidad de la base externa
Freqe = pd.to_numeric(
    GymPulse["Frecuencia_Cancelaciones_Dia"],
    errors="coerce"
)

Seve = pd.to_numeric(
    GymPulse["Severidad_COP_Cliente"],
    errors="coerce"
)

# Distribución agregada de pérdidas externa
LDAe = (Freqe * Seve).dropna().to_numpy(dtype=float)

print("Cantidad de datos externos:", len(LDAe))
print("Media de los datos externos:", np.mean(LDAe))
print("Asimetría de los datos externos:", pd.Series(LDAe).skew())
print("Curtosis de los datos externos:", pd.Series(LDAe).kurt())

plt.figure(figsize=(8,5))
sns.kdeplot(LDAe)
plt.title("Distribución agregada de pérdidas - GymPulse")
plt.xlabel("Pérdida económica diaria")
plt.ylabel("Densidad")
plt.grid()
plt.show()
```

## 10.4 Integración de los datos externos

Se utiliza un **umbral de pertenencia del 90%**, igual que en el ejercicio de integración trabajado en clase.  
Cuando un dato externo presenta una pertenencia superior a 0,90 frente a alguno de los concentradores internos, se integra a la caja correspondiente.


```python
# Crear copias para conservar intactas las cajas internas originales
Boxes_integradas = [list(caja) for caja in Boxes_int]
XC_integrado = XC_int.copy()

# Contador de datos externos integrados
m1 = 0

for k in range(len(LDAe)):

    # Evitar división entre cero
    denominador = np.where(
        np.abs(XC_integrado) < 1e-12,
        1e-12,
        np.abs(XC_integrado)
    )

    # Valores de pertenencia del dato externo
    VP = np.exp(
        -0.5 * ((XC_integrado - LDAe[k]) / denominador)**2
    )

    # Integrar si el mayor valor de pertenencia supera el 90%
    if np.max(VP) > 0.90:

        m1 = m1 + 1

        # Caja de mayor pertenencia
        nc = np.argmax(VP)

        # Agregar dato externo
        Boxes_integradas[nc].append(LDAe[k])

        # Actualizar concentrador
        XC_integrado[nc] = (
            XC_integrado[nc] + LDAe[k]
        ) / 2

print("Los datos externos integrados fueron:", m1)
print(
    "Porcentaje de datos externos integrados:",
    round((m1 / len(LDAe)) * 100, 2),
    "%"
)

print("Concentradores después de integrar GymPulse:")
print(XC_integrado)
```

## 10.5 Muestreo después de integrar GymPulse


```python
LDA_muestra_integrada = []

random.seed(123)

for j in range(5):

    LDA_muestra_integrada.extend(
        random.choices(
            Boxes_integradas[j],
            k=pm_int[j]
        )
    )

LDA_muestra_integrada = np.array(LDA_muestra_integrada)

print(
    "Cantidad de datos de la muestra después de la integración:",
    len(LDA_muestra_integrada)
)
```

## 10.6 Comparación de las tres distribuciones


```python
plt.figure(figsize=(10,6))

sns.kdeplot(
    LDA_int,
    label="Datos originales FitBalance"
)

sns.kdeplot(
    LDA_muestra_interna,
    label="Datos muestreados"
)

sns.kdeplot(
    LDA_muestra_integrada,
    label="Datos muestreados + GymPulse"
)

plt.title("Comparación de distribuciones antes y después de la integración")
plt.xlabel("Pérdida económica diaria")
plt.ylabel("Densidad")
plt.legend()
plt.grid()
plt.show()
```

## 10.7 Caracterización estadística antes y después de la integración


```python
# Medidas usando la misma función creada anteriormente en el reto
hist_orig_int, medidas_orig_int = MedidasDTC(LDA_int, NI=10)

hist_muestra_int, medidas_muestra_int = MedidasDTC(
    LDA_muestra_interna,
    NI=10
)

hist_integrada_int, medidas_integrada_int = MedidasDTC(
    LDA_muestra_integrada,
    NI=10
)

comparacion_integracion = pd.concat(
    [
        medidas_orig_int.rename("Original FitBalance"),
        medidas_muestra_int.rename("Muestra Interna"),
        medidas_integrada_int.rename("Muestra + GymPulse")
    ],
    axis=1
)

display(comparacion_integracion)
```

## 10.8 Discrepancias respecto a la distribución original


```python
errores_integracion = pd.DataFrame(index=comparacion_integracion.index)

errores_integracion["Error % Muestra Interna"] = (
    np.abs(
        comparacion_integracion["Muestra Interna"]
        - comparacion_integracion["Original FitBalance"]
    )
    / np.abs(comparacion_integracion["Original FitBalance"])
) * 100

errores_integracion["Error % Muestra + GymPulse"] = (
    np.abs(
        comparacion_integracion["Muestra + GymPulse"]
        - comparacion_integracion["Original FitBalance"]
    )
    / np.abs(comparacion_integracion["Original FitBalance"])
) * 100

display(errores_integracion)
```

## Análisis de resultados de la integración

El criterio utilizado para decidir la integración fue un valor de pertenencia superior al 90%, de manera que únicamente se incorporaran observaciones externas con un comportamiento suficientemente cercano a las categorías de riesgo de la base interna.

Con los datos suministrados, aproximadamente la mitad de las observaciones de GymPulse cumplen con este criterio de pertenencia. Esto demuestra que existe una parte importante de la información externa que presenta características compatibles con la distribución de pérdidas de FitBalance, pero también existe otro grupo de observaciones que no presenta una afinidad suficiente y, por lo tanto, no se integra.

La comparación de las distribuciones permite analizar el efecto que produce la incorporación de información externa sobre la estructura original de las pérdidas. La muestra obtenida únicamente con los datos internos mantiene una mayor cercanía con la distribución original, mientras que la inclusión de GymPulse modifica principalmente el nivel promedio y la dispersión de las pérdidas. Esto ocurre porque la base externa presenta, en términos generales, pérdidas de menor magnitud que las observadas originalmente en FitBalance.

Por esta razón, el criterio de pertenencia cumple una función importante: evita integrar automáticamente toda la base externa y permite seleccionar únicamente aquellos registros que presentan mayor similitud con las categorías de riesgo internas. Si se quisiera obtener una integración más estricta y conservar todavía más la estructura de la distribución original, sería posible aumentar el umbral de pertenencia, por ejemplo del 90% al 95%, de forma similar a la recomendación realizada en el ejercicio de integración anterior.

En conclusión, la integración con GymPulse permite ampliar la información disponible para el análisis del riesgo, pero evidencia que la calidad de la integración depende de la afinidad existente entre los datos internos y externos. Por esta razón, además de aumentar el volumen de información, es necesario evaluar cómo la incorporación de nuevos registros modifica la media, la dispersión, la asimetría y la curtosis de la distribución original.
