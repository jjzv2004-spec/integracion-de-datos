3. Se procede con la construccion del histograma de frecuencias

```python
NI = 10 # Numero INtervalos
counts, Limits = np.histogram(LDA,bins= NI)# Agrupar los datos
LI= Limits[:-1] # limite inferior
LS= Limits[1:] # limite superior
fi=counts/sum(counts) # POrcentaje de datos por intervalo
MC= (LI+LS)/2 # Marca de clase

# Create a new DataFrame with the histogram data
histogram_data = pd.DataFrame({
    'LI': LI,
    'LS': LS,
    'ND': counts,
    'fi': fi,
    'MC': MC
})
display(histogram_data)
```

3. Medias de tendencia Central

```python
np.set_printoptions(suppress=True)
#Medida
U=np.sum(MC*fi)
print(f" La media de las perdidas es : {U:.2f}")

#Varianza
varc=np.sum((MC-U)**2 * fi) # Corrected variance calculation
print(f" La varianza es : {varc:.2f}")

# INtervalo de variacion de las perdidas
sigma=np.sqrt(varc)
print(f" El limite inferior es: {U-2*sigma:.2f}")
print(f" El limite superior es: {U+2*sigma:.2f}")

#Coefieciente de asimetria
cas=np.sum(((MC-U)**3)*fi)/((sigma**3))
print(f" El coeficiente de asimetria es: {cas:.2f}")

#Coeficiente de curtosis
cur=np.sum(((MC-U)**4)*fi)/((sigma**4))
print(f" El coeficiente de curtosis es: {cur:.2f}")
```

```python
def MedidasDTC(Xi, NI):

    # NI: Número de intervalos en los cuales quiero agrupar los datos
    # Xi: Indica una variable aleatoria de analiss

    counts, limits = np.histogram(Xi, bins=NI)   # Agrupemos los datos en 10 intervalos

    LI = limits[:-1]          # Límites inferiores del histograma
    LS = limits[1:]           # Límites superiores del histograma

    fi = counts / sum(counts)     # Porcentaje de datos por intervalo

    MC = (LI + LS) / 2            # Marca de Clase: Dato representativo de cada intervalo

    df = pd.DataFrame(
        np.column_stack((LI, LS, counts, fi, MC))
    )

    df.columns = ["LI", "LS", "ND", "fi", "MC"]

    display(df)

    # 1. Calculamos la media
    u = np.sum(MC * fi)

    # 2. Varianza: Indica qué tan dispersos están los datos alrededor de la Media
    varc = np.sum(((MC - u) ** 2) * fi)

    # 3. Calculamos la Desviación
    sigma = np.sqrt(varc)

    # 4. Coeficiente de asimetría
    cas = np.sum(((MC - u) ** 3) * fi) / (sigma ** 3)

    return df, u, varc, sigma, cas
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display

# Leer archivo
XDB = pd.read_excel(path, sheet_name=0)

# Variables
Freq = XDB.iloc[:, 3]
Sev = XDB.iloc[:, 4]
LDA = Freq * Sev

# ==========================================
# FRECUENCIA
# ==========================================

print("========== FRECUENCIA ==========")

df, uf, varf, desvf, casf = MedidasDTC(Freq, 10)

print("Media:", uf)
print("Varianza:", varf)
print("Desviación estándar:", desvf)
print("Coeficiente de asimetría:", casf)

display(df)

# ==========================================
# SEVERIDAD
# ==========================================

print("\n========== SEVERIDAD ==========")

dfs, us, vars, desvs, cass = MedidasDTC(Sev, 10)

print("Media:", us)
print("Varianza:", vars)
print("Desviación estándar:", desvs)
print("Coeficiente de asimetría:", cass)

display(dfs)

# ==========================================
# DISTRIBUCIÓN AGREGADA (LDA)
# ==========================================

print("\n========== DISTRIBUCIÓN AGREGADA ==========")

dfLDA, uLDA, varLDA, desvLDA, casLDA = MedidasDTC(LDA, 10)

print("Media:", uLDA)
print("Varianza:", varLDA)
print("Desviación estándar:", desvLDA)
print("Coeficiente de asimetría:", casLDA)

display(dfLDA)

# ==========================================
# HISTOGRAMA
# ==========================================

plt.figure(figsize=(8,5))

sns.histplot(LDA,
             bins=10,
             kde=True,
             color='blue')

plt.title("Distribución Agregada de las Pérdidas - EVERGREEN")
plt.xlabel("Pérdidas")
plt.ylabel("Frecuencia")
plt.grid(True)

plt.show()
```
